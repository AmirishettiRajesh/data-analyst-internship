The next cell demonstrates variables, operators, conditions, loops, functions, built-in collections, a handled conversion error, and context-managed CSV file handling.

```python
student_count = len(data)
study_hours = 5.8
assessment_name = "Data Science Fundamentals"
is_reproducible = True
converted_count = int("20")
arithmetic_result = 8 + 2 * 3
comparison_result = student_count >= converted_count
logical_result = is_reproducible and study_hours > 0

if study_hours >= 8:
    study_band = "high"
elif study_hours >= 4:
    study_band = "moderate"
else:
    study_band = "low"

study_totals = []
for hour_count in [2, 4, 6]:
    study_totals.append(hour_count * 2)
countdown = []
counter = 3
while counter > 0:
    countdown.append(counter)
    counter -= 1

def pass_rate(scores, pass_mark=70):
    """Return the proportion of scores meeting a pass mark."""
    if not scores:
        return 0.0
    return sum(score >= pass_mark for score in scores) / len(scores)

try:
    invalid_integer = int("not numeric")
except ValueError:
    invalid_integer = None

# Context managers close the file; remove the temporary example after reading it.
example_path = OUTPUT_DIR / "file_handling_example.csv"
try:
    with example_path.open("w", encoding="utf-8") as output_file:
        output_file.write("concept,value\nassessment,Data Science Fundamentals\n")
    with example_path.open("r", encoding="utf-8") as input_file:
        first_line = input_file.readline().strip()
finally:
    example_path.unlink(missing_ok=True)
print({"study_band": study_band, "arithmetic": arithmetic_result,
       "comparison": comparison_result, "logical": logical_result,
       "for_loop": study_totals, "while_loop": countdown,
       "pass_rate": pass_rate([68, 75, 91]), "handled_error": invalid_integer,
       "file_header": first_line})
```

## 9. Pandas

Load the root CSV into a DataFrame, inspect head/tail, shape, schema, and descriptive statistics, then select, filter, sort, group, and aggregate. Check missing values and duplicate rows before cleaning.

```python
data = pd.read_csv(CSV_PATH)
validate_data(data)
print("Shape:", data.shape)
print("DataFrame info:")
data.info()
display(data.head())
display(data.tail())
display(data.describe().round(2))

selected_columns = data[["Name", "Study_Hours", "Exam_Score"]]
high_attendance = data.loc[data["Attendance"] >= 90, ["Student_ID", "Name", "Attendance"]]
sorted_students = data.sort_values("Exam_Score", ascending=False)
gender_summary = data.groupby("Gender").agg(
    Student_Count=("Student_ID", "count"),
    Mean_Exam_Score=("Exam_Score", "mean"),
    Median_Exam_Score=("Exam_Score", "median"),
)
print("Selected columns:\n", selected_columns.head().to_string(index=False))
print("Rows with attendance >= 90:", len(high_attendance))
print("Top exam scores:\n", sorted_students[["Name", "Exam_Score"]].head().to_string(index=False))
display(gender_summary.round(2))
print("Missing values by column:\n", data.isna().sum().to_string())
print("Duplicate rows:", int(data.duplicated().sum()))
```

## 10. Data Cleaning

The source dataset is validated as complete. To demonstrate how decisions are made transparently, create a separate copy with one missing Attendance value and one repeated row. Detect both, remove the duplicate, impute attendance with the copy's median, and flag potential exam-score outliers using the 1.5 x IQR rule. This does not silently alter the source.

```python
imperfect_data = data.copy()
imperfect_data.loc[0, "Attendance"] = np.nan
imperfect_data = pd.concat([imperfect_data, imperfect_data.iloc[[1]]], ignore_index=True)
missing_before = int(imperfect_data.isna().sum().sum())
duplicates_before = int(imperfect_data.duplicated().sum())
cleaned_data = imperfect_data.drop_duplicates().copy()
attendance_median = cleaned_data["Attendance"].median()
cleaned_data["Attendance"] = cleaned_data["Attendance"].fillna(attendance_median)
first_quartile = cleaned_data["Exam_Score"].quantile(0.25)
third_quartile = cleaned_data["Exam_Score"].quantile(0.75)
interquartile_range = third_quartile - first_quartile
lower_bound = first_quartile - 1.5 * interquartile_range
upper_bound = third_quartile + 1.5 * interquartile_range
outlier_count = int((~cleaned_data["Exam_Score"].between(lower_bound, upper_bound)).sum())
quality_results = demonstrate_data_quality(data)
print({"missing_before": missing_before, "duplicates_before": duplicates_before,
       "missing_after": int(cleaned_data.isna().sum().sum()),
       "duplicates_after": int(cleaned_data.duplicated().sum()),
       "attendance_median_used": attendance_median, "exam_score_outliers": outlier_count})
validate_data(data)
```

## 13. Best Practices

- Reproducibility: keep source data, package requirements, and random seeds explicit.
- Validation: check schemas, types, allowed ranges, nulls, duplicates, and potential outliers before analysis.
- Readability: use meaningful names, comments for non-obvious decisions, and documentation for reusable functions.
- Privacy: use fictional or appropriately de-identified information; do not expose student-level records.
- Visualization: label units, choose honest scales, and avoid implying a trend from unordered records.
- Interpretation: state sample size and limitations; correlation alone does not support causal conclusions.
